# 01 - Exploratory Data Analysis

Loads the original UCI dataset, checks its structure, and generates meaningful plots in `outputs/`.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from preprocessing import load_hourly_data, validate_dataset
from eda import run_eda
df = load_hourly_data(ROOT / 'dataset' / 'hour.csv')
validate_dataset(df)
print('Shape:', df.shape)
display(df.head())
display(df.dtypes.to_frame('dtype'))
display(df.describe().T)
display(df.isna().sum().to_frame('missing'))
print('Duplicate rows:', df.duplicated().sum())
findings = run_eda(df, ROOT / 'outputs')
print('Measured findings:', findings)
print('Interpretation: plots are saved with titles and axis labels; outliers are investigated rather than automatically removed because high demand may be genuine.')
```
